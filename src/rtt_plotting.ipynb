```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
df = pd.read_csv('~/thunderbots/it_androids.csv')
# df = pd.read_csv('~/thunderbots/rione_final_rtt.csv')
df = df[df['rtt'] < 1e5]
df['rtt'] = df['rtt'] * 1000
start = 20000
df = df[start:start+1000]

# Convert the timestamp column to datetime format
# df['time'] = pd.to_datetime(df['time'])

# Plot the data
plt.figure(figsize=(20,6))
plt.plot(df['time'], df['rtt'], label='RTT over time')
plt.xlabel('Time')
plt.ylabel('RTT in seconds')
plt.title('RTT over Time')
plt.legend()
plt.show()
```
