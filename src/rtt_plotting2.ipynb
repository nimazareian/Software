```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.float_format', lambda x: '%.3f' % x)

# Read the CSV file
df = pd.read_csv('~/thunderbots/rione_final_rtt.csv')
df['rtt'] = 1000 * df['rtt']

# df.iloc[[2831]]
# df.head()

# Calculate percentiles
# df['rtt'] = pd.to_numeric(df['rtt'])
percentiles = df['rtt'].describe(percentiles=[.50, .80, .90, .95, .98, .99])

# Print percentiles
print("Percentiles:")
print(percentiles)

# Count data points
data_points = df['rtt'].count()

# Print number of data points
print("\nNumber of data points:", data_points)
```
